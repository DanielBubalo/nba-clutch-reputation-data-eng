```python
def adv_stats_table() -> pd.DataFrame:
    adv_stats = leaguedashplayerstats.LeagueDashPlayerStats(season="2025-26", 
                                                            measure_type_detailed_defense="Advanced",
                                                            season_type_all_star="Regular Season")
    adv_stats_df = adv_stats.get_data_frames()[0]
    return adv_stats_df.columns

adv_stats_table()
```
